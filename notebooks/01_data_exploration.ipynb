```python
#Import libraries
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries loaded!")
```

```python
# Load the dataset
dataset = load_dataset("mteb/NepaliNewsClassification")
print(dataset)
```

```python
#Look at first few examples
for i in range(3):
    example = dataset['train'][i]
    print(f"Example {i+1}:")
    print(f"Label: {example['label']}")
    print(f"Text: {example['text'][:100]}...")
    print()
```

```python
#Count samples
print(f"Training samples: {len(dataset['train'])}")
print(f"Test samples: {len(dataset['test'])}")
```

```python
#Check class distribution
label_counts = pd.Series(dataset['train']['label']).value_counts()
print(label_counts)
print(f"\nPercentage:")
print((label_counts / len(dataset['train']) * 100).round(2))
```

```python
#Check text lengths
lengths = [len(text) for text in dataset['train']['text']]
print(f"Average length: {sum(lengths)/len(lengths):.0f} characters")
print(f"Min: {min(lengths)}, Max: {max(lengths)}")
```
